# Diamond 50 cm Headwind Level2 — single-run pilot
Uses the existing Bideal calibration and removes the initial unchanged-SOC waiting interval.
Reference stages are 100–82, 82–52 and 52–20%. The reference starting SOC is numerically aligned to the first retained observed SOC.
P3–P5 High is completed using the fitted Medium relative drain factor; P1 Low and P2 High have short support. These assumptions are recorded in manifest.json. This is a pilot, not active training data.
Only the common five-drone hover interval is used. No aircraft controllers are imported or called.


```python
from pathlib import Path
import json
import pandas as pd
from IPython.display import display, Image
trial = Path.cwd()
if not (trial / "manifest.json").exists():
    raise RuntimeError("Open this notebook with its containing trial directory as the working directory")
manifest = json.loads((trial / "manifest.json").read_text())
display(pd.read_csv(trial / "stage_rates.csv"))
display(pd.read_csv(trial / "initial_plateau_and_window_audit.csv"))
display(Image(filename=str(trial / "observations_and_fit.png")))
```
